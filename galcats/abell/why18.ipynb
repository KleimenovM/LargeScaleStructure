```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import healpy as hp
```

```python
data = pd.read_csv("WHY18table1.dat", delimiter=' ', index_col=False)
data = data[data["R_L"] > 150]

data
```

```python
nside=128
npix = hp.nside2npix(nside)
map = np.zeros(npix)

hp.mollview(map, cmap='RdBu', min=-1, max=1, cbar=False, coord='G')
hp.projscatter(np.pi/2 - np.deg2rad(data["Dec"]), np.deg2rad(data["RA"]), c=data["z"], coord='C', s=20)
hp.graticule()
```
