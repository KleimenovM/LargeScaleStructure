```python
import numpy as np
import pandas as pd
```

```python
# for column description see https://glade.elte.hu/
cols_to_keep = [8, 9, 28]

reader = pd.read_csv(
    "GLADE+.txt",
    delimiter=' ',   # or sep=","
    header=None,
    usecols=cols_to_keep,
    chunksize=2_500_000    # 22'5 10^6 mil galaxies
)

for i, chunk in enumerate(reader):
    chunk_clean = chunk[cols_to_keep]
    chunk_clean.to_csv(f"glade_regroup/GLADE_{i+1}_z.csv", header=False, index=False)
```
