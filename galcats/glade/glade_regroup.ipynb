```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
below = []
for i in range(1, 11):
    df = pd.read_csv(f"glade_regroup/GLADE_{i}_z.csv", names = ["ra", "dec", "z"])
    df_below_i = df[df["z"] < 0.1]
    below_i = df_below_i.to_numpy(dtype=np.float32)
    below.append(below_i) 
```

```python
below_np = np.concatenate(below)
print(below_np.shape)
below_np
```

```python
fname = "GLADE_zb01.npz"
np.savez(fname, ra=below_np[:, 0], dec=below_np[:, 1], z=below_np[:, 2])
```
